## Question 10 – Student Ranking

```python
import numpy as np
marks = np.array([245, 278, 219, 290, 256])
sorted_indices = np.argsort(marks)[::-1]
sorted_marks = marks[sorted_indices]
print("Original Marks:", marks)
print("Marks in descending order:")
print(sorted_marks)
print("Ranking:")
for rank, index in enumerate(sorted_indices, start=1):
    print("Rank", rank, "- Student", index + 1, "- Marks:", marks[index])
```

## Question 12 – Missing Marks

## Question 13 – Grade Classification

```python
import numpy as np
marks = np.array([95, 82, 74, 61, 45])
grades = np.select(
    [
        marks >= 90,
        marks >= 80,
        marks >= 70,
        marks >= 60
    ],
    [
        "A",
        "B",
        "C",
        "D"
    ],
    default="F"
)

print("Marks:", marks)
print("Grades:", grades)
```

## Question 14 – Random Marks Generation

```python
import numpy as np
np.random.seed(10)
marks = np.random.randint(0, 101, 5)
print("Randomly Generated Marks:")
print(marks)
print("Total:", np.sum(marks))
print("Average:", np.mean(marks))
print("Highest:", np.max(marks))
print("Lowest:", np.min(marks))
print("Standard Deviation:", np.std(marks))
```

## Question 15 – Student Performance Analysis

```python
import numpy as np
marks = np.array([
    [78, 85, 90],
    [65, 72, 80],
    [88, 91, 84],
    [56, 62, 70],
    [95, 89, 92]
])
total = np.sum(marks, axis=1)
average = np.mean(marks, axis=1)
highest = np.max(marks, axis=1)
lowest = np.min(marks, axis=1)
class_average = np.mean(marks)
above_average = average > class_average
print("Student Marks:")
print(marks)
print("Total Marks:")
print(total)
print("Average Marks:")
print(average)
print("Highest Marks:")
print(highest)
print("Lowest Marks:")
print(lowest)
print("Class Average:", class_average)
print("Students above class average:")

for i in range(len(marks)):
    if above_average[i]:
        print("Student", i + 1, "- Average:", average[i])
```
