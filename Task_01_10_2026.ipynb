```python
#Online Shopping

discount=list(map(int,input().split(",")))
maxi=0
sumi=0
for i in range(len(discount)):
  sumi+=discount[i]
  if sumi<0:
    sumi=0
  maxi=max(sumi,maxi)
print("Maximum Discount:",maxi)
```

```python
#Bank Transaction
trans=list(map(int,input().split(",")))
tar=int(input())
cnt=0
for i in range(len(trans)):
  total=0
  for j in range(i,len(trans)):
    total+=trans[j]
    if total==tar:
      cnt+=1
print("Count:",cnt)
```

```python
#Password Security Analysis
password=input()
longest=""
for i in range(len(password)):
  for j in range(i,len(password)):
    sub=password[i:j+1]
    if sub==sub[::-1] and len(sub)>len(longest):
      longest=sub
print("Longest Palindromic String:",longest)
```
